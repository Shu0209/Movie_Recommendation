```python
import nltk
import re
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
```

```python
def preprocess(text):
    # lowercase
    text=str(text).lower()
    # remove puncuation
    text=re.sub(r'[^a-zA-Z\s]','',text)
    # Tokenization
    words=word_tokenize(text)
    # Remove Stopwords
    words=[word for word in words if word not in stop_words]
    # Lemmatization
    words=[lemmatizer.lemmatize(word) for word in words]
    return " ".join(words)
    
```

```python
df['tag1'][0]
```

```python
preprocess(df['tag1'][0])
```

```python
df["tag1"]=df['tag1'].apply(preprocess)
df['tag2']=df['tag2'].apply(preprocess)
```

```python
# Vector Embedding
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf=TfidfVectorizer(max_features=50000,ngram_range=(1,2))
```

```python
tfidf_metrix_tag1=tfidf.fit_transform(df['tag1'])
tfidf_metrix_tag2=tfidf.fit_transform(df['tag2'])
```

```python
tfidf_metrix_tag1.toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title,n=10):       #n=10 means 10 movies suggestions
    if title not in indices:
        return ["Movie not found"]
    idx=indices[title]
    score1=cosine_similarity(tfidf_metrix_tag1[idx],tfidf_metrix_tag1).flatten()
    score2=cosine_similarity(tfidf_metrix_tag2[idx],tfidf_metrix_tag2).flatten()
    score=(score1+score2)/2
    similarity_idx=score.argsort()[::-1][1:n+1]
    return df['title'].iloc[similarity_idx]
```

```python
recommend('Inception')
```
